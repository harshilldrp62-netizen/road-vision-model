```python
import torch

class YOLOWrapper(torch.nn.Module):
    def __init__(self, yolo_model):
        super().__init__()
        self.model = yolo_model

    def forward(self, x):
        output = self.model(x)
        if isinstance(output, (tuple, list)):
            return output[0]   # take just the main prediction tensor
        return output
```

```python
from pytorch_grad_cam import EigenCAM
from pytorch_grad_cam.utils.image import show_cam_on_image
import cv2
import numpy as np
import torch

def generate_heatmap(image_path, model):
    img = cv2.imread(image_path)
    if img is None:
        raise FileNotFoundError(f"Could not read image at: {image_path}")

    # Wrap the underlying torch model
    wrapped_model = YOLOWrapper(model.model)
    wrapped_model.eval()

    target_layers = [model.model.model[-2]]
    cam = EigenCAM(wrapped_model, target_layers)

    img_resized = cv2.resize(img, (640, 640))
    rgb_img = img_resized[:, :, ::-1] / 255.0

    device = next(model.model.parameters()).device
    input_tensor = torch.from_numpy(rgb_img).permute(2, 0, 1).unsqueeze(0).float().to(device)

    grayscale_cam = cam(input_tensor)[0]
    cam_image = show_cam_on_image(rgb_img.astype(np.float32), grayscale_cam, use_rgb=True)
    return cam_image
```

```python
import os

demo_images = [
    "PotHoles_plus_unpaved_178.jpg",
    "PotHoles_plus_unpaved_46.jpg",
    "SB__93.jpg",
    "SB__217.jpg",
    "Speed_91.jpg",
    'Speed_177.jpg',
    "UngradedRoad_138.jpg",
    "UngradedRoad_35.jpg",
    "UnPavedRoad__203.jpg",
    "UnPavedRoad__87.jpg"



]

os.makedirs("demo_output", exist_ok=True)

for img_name in demo_images:
    path = os.path.join("data/images/test", img_name)
    try:
        heatmap = generate_heatmap(path, model)
        cv2.imwrite(f"demo_output/heatmap_{img_name}", heatmap)
        print(f"Done: {img_name}")
    except Exception as e:
        print(f"Failed on {img_name}: {e}")
```

```python
TYPE_WEIGHT = {"pothole": 1.0, "unpaved_road": 0.7, "speed_breaker": 0.3}

def compute_severity(damage_class, confidence, bbox, img_w, img_h):
    x1, y1, x2, y2 = bbox
    area_ratio = ((x2-x1)*(y2-y1)) / (img_w * img_h)
    raw_score = 0.5*min(area_ratio*5, 1.0) + 0.2*confidence + 0.3*TYPE_WEIGHT.get(damage_class, 0.5)
    if raw_score < 0.35: severity = "Low"
    elif raw_score < 0.55: severity = "Medium"
    elif raw_score < 0.75: severity = "High"
    else: severity = "Critical"
    return severity, round(raw_score, 3)

def generate_explanation(damage_class, confidence, area_ratio, severity):
    descriptions = {
        "pothole": "a depression in the road surface that can damage vehicles",
        "unpaved_road": "a stretch of road lacking proper paving, causing dust and instability",
        "speed_breaker": "a raised road hazard that may be unmarked or poorly visible",
    }
    return (f"Detected {damage_class.replace('_',' ')} with {confidence*100:.1f}% confidence, "
            f"covering ~{area_ratio*100:.1f}% of the visible area. {descriptions[damage_class]}. "
            f"Classified as {severity} severity.")

# Run on your demo images, print/save results for each detection
for img_name in demo_images:
    path = os.path.join("data/images/test", img_name)
    results = model(path, conf=0.4)[0]
    img_h, img_w = results.orig_shape
    for box in results.boxes:
        cls_name = model.names[int(box.cls)]
        conf = float(box.conf)
        x1, y1, x2, y2 = box.xyxy[0].tolist()
        area_ratio = ((x2-x1)*(y2-y1)) / (img_w*img_h)
        severity, score = compute_severity(cls_name, conf, (x1,y1,x2,y2), img_w, img_h)
        explanation = generate_explanation(cls_name, conf, area_ratio, severity)
        print(f"{img_name}: {explanation}")
```
